## DEEP LEARNING DAY - 1 : Artificial Neural Network

### What is Deep Learning?

**Deep Learning** is a subset of **Machine Learning (ML)** that uses **artificial neural networks** with many layers (hence “deep”) to automatically learn features and patterns from data.

#### Key Characteristics:

* Uses **neural networks** with multiple layers (input, hidden, output).
* Learns **complex patterns** directly from raw data (images, text, etc.).
* Requires **large datasets** and **high computational power** (GPUs/TPUs).
* Performs **automatic feature extraction** — no need for manual feature engineering.

---

### How is Deep Learning different from Machine Learning?

| Feature                          | Machine Learning (ML)                         | Deep Learning (DL)                                |
| -------------------------------- | --------------------------------------------- | ------------------------------------------------- |
| **Feature Engineering**          | Manual (human-defined features)               | Automatic (learned from raw data)                 |
| **Data Dependency**              | Works well with small to medium datasets      | Needs large datasets to perform well              |
| **Model Type**                   | Algorithms like SVM, Decision Tree, KNN, etc. | Multi-layer Neural Networks (ANN, CNN, RNN, etc.) |
| **Training Time**                | Faster                                        | Slower (depends on model depth and data size)     |
| **Hardware Needs**               | Can run on CPU                                | Needs GPU/TPU for efficient training              |
| **Interpretability**             | More interpretable (e.g., Decision Trees)     | Less interpretable (Black-box models)             |
| **Performance on Complex Tasks** | Limited                                       | Excels in tasks like vision, speech, NLP          |

---

### In Summary:

* **ML** = good for structured data and simpler tasks.
* **DL** = best for unstructured data like images, audio, and text where learning deep hierarchical features is essential.

### ✅ What Are Activation Functions?

**Activation functions** are mathematical operations applied to the output of a neuron. They decide **whether a neuron should “fire” or not**, and introduce **non-linearity** into neural networks — allowing them to **learn complex patterns** beyond linear relationships.

---

### 🧠 Why Activation Functions Matter

1. **Non-Linearity**
   Without activation functions, neural networks would behave like a simple linear model regardless of depth.

2. **Decision Making**
   They control how much signal should pass to the next layer.

3. **Helps in Convergence**
   Different activation functions help in faster and more stable training.

---

### 🔢 Common Activation Functions

---

#### 1. **Step Function**

* Used in perceptrons.
* Output:

  $$
  f(x) = 
  \begin{cases}
    1 & \text{if } x \geq 0 \\
    0 & \text{if } x < 0
  \end{cases}
  $$
* ❌ **Problem**: Not differentiable → not suitable for backpropagation.

---

#### 2. **Sigmoid Function**

* Formula:

  $$
  f(x) = \frac{1}{1 + e^{-x}}
  $$
* Output: (0, 1)
* ✅ Good for probabilities
* ❌ **Vanishing gradient problem** when $x$ is large/small.

---

#### 3. **Tanh (Hyperbolic Tangent)**

* Formula:

  $$
  f(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}
  $$
* Output: (-1, 1)
* ✅ Zero-centered
* ❌ Still suffers from vanishing gradients

---

#### 4. **ReLU (Rectified Linear Unit)**

* Formula:

  $$
  f(x) = \max(0, x)
  $$
* ✅ Most widely used due to simplicity and efficiency
* ✅ Avoids vanishing gradient (for positive values)
* ❌ **Dying ReLU**: Neurons can “die” if inputs are always negative

---

#### 5. **Leaky ReLU**

* Formula:

  $$
  f(x) = 
  \begin{cases}
    x & \text{if } x > 0 \\
    \alpha x & \text{if } x \leq 0 \ (\alpha \approx 0.01)
  \end{cases}
  $$
* ✅ Fixes Dying ReLU by allowing a small slope for negative inputs.

---

#### 6. **Parametric ReLU (PReLU)**

* Like Leaky ReLU but $\alpha$ is learned during training.
* ✅ Adaptive and more flexible

---

#### 7. **ELU (Exponential Linear Unit)**

* Formula:

  $$
  f(x) = 
  \begin{cases}
    x & \text{if } x \geq 0 \\
    \alpha (e^x - 1) & \text{if } x < 0
  \end{cases}
  $$
* ✅ Smooth output, better learning for negative inputs

---

#### 8. **Softmax**

* Converts a vector of raw scores into probabilities.
* Formula:

  $$
  \text{Softmax}(x_i) = \frac{e^{x_i}}{\sum_{j=1}^{n} e^{x_j}}
  $$
* ✅ Used in **output layers of multi-class classification**

---

### 🟨 Summary Table

| Activation Function | Output Range | Use Case                     | Issues                     |
| ------------------- | ------------ | ---------------------------- | -------------------------- |
| Step                | {0, 1}       | Perceptrons                  | Not differentiable         |
| Sigmoid             | (0, 1)       | Binary classification        | Vanishing gradients        |
| Tanh                | (-1, 1)      | Hidden layers                | Vanishing gradients        |
| ReLU                | \[0, ∞)      | Most hidden layers           | Dying neurons              |
| Leaky ReLU          | (-∞, ∞)      | Fix for ReLU                 | Small gradient in negative |
| PReLU               | (-∞, ∞)      | Learnable leaky slope        | Slightly complex           |
| ELU                 | (-α, ∞)      | Better for noisy data        | Slightly slower            |
| Softmax             | (0, 1) sum=1 | Output layer for multi-class | Not for hidden layers      |

## Part - 1 : Data Preprocessing 

### Importing the Dataset

```python
df = pd.read_csv('../datasets/Churn_Modelling.csv')
df
```

```python
X = df.iloc[: , 3:-1].values
y = df.iloc[:, -1].values
print(X)
print('\n')
print(y)
```

### Encoding the Categorical Data

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X[:, 2] = le.fit_transform(X[:, 2])
```

```python
print(X)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers = [('encoder', OneHotEncoder(), [1])], remainder = 'passthrough')
X = np.array(ct.fit_transform(X))
```

```python
print(X)
```

### Splitting the dataset into training and test set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
print(X_train)
print('\n')
print(X_test)
print('\n')
print(y_train)
print('\n')
print(y_test)
```

### Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

## Part - 2 : Building ANN

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
ann = Sequential([
    Dense(units = 6, activation = 'relu'),
    Dense(units = 6, activation = 'relu'),
    Dense(units = 1, activation = 'sigmoid')
])
```

### Compiling the ANN

```python
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

### Training the ANN on the Training set

```python
ann.fit(X_train, y_train, batch_size = 32, epochs = 100)
```

```python
y_pred = ann.predict(X_test)
y_pred = (y_pred > 0.5)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```
