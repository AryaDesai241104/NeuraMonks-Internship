# Reccurent Neural Network

### What is an RNN?

A **Recurrent Neural Network (RNN)** is a type of artificial neural network that is specially designed to handle sequential data, such as time series, natural language, or audio. Unlike traditional feedforward networks, where all inputs are independent, RNNs excel at capturing dependencies across time by maintaining an internal memory of previous inputs.

#### How RNNs Work

-   **Sequence Awareness**: RNNs process an input sequence one element at a time while retaining context from earlier elements via a "hidden state."
-   **Information Feedback**: At each step, the network receives the current input and the previous hidden state, processes them, and outputs both a result and a new hidden state for the next step.
-   **Unrolling**: For training, RNNs are "unrolled" for a series of time steps, forming a deep network over time.

#### Mathematical Representation

Given input sequence $x_t$, the RNN updates as follows:
-   $h_t = f(W_{hh} \cdot h_{t-1} + W_{xh} \cdot x_t + b_h)$
-   $y_t = g(W_{hy} \cdot h_t + b_y)$
where $h_t$ is the hidden state at time $t$, and $y_t$ is the output.

#### Key Characteristics

-   **Memory**: The hidden state acts as RNN's "memory," retaining previous information.
-   **Variable-Length Input/Output**: RNNs can handle sequences of varying lengths.
-   **Backpropagation Through Time (BPTT)**: The training process involves extending standard backpropagation across all time steps (unrolled network).\[1]

#### Challenges

-   **Vanishing and Exploding Gradients**: When learning long-term dependencies, gradients can shrink or grow uncontrollably, making training unstable or ineffective.
-   **Limited Long-Term Memory**: Classic RNNs struggle with retaining long-range context due to the above gradient problem.\[4]\[2]

### Applications

-   Language modeling
-   Speech recognition
-   Machine translation
-   Sentiment analysis
-   Stock prediction

### 2. Long Short-Term Memory (LSTM)

#### Why LSTM Was Invented

RNNs are powerful but typically ineffective at capturing long-range dependencies because of vanishing gradients. **LSTM** networks were designed specifically to address this limitation by introducing a more sophisticated memory structure.

#### LSTM Architecture

An LSTM unit is made up of a "cell state" and three main gates:

-   **Cell State**: The "memory" of the network, designed to carry relevant long-term information across many time steps.
-   **Gates**: These are structures that control the flow of information. Each gate is based on a sigmoid activation function (values between 0 and 1) multiplied by the relevant values.

#### 1. Forget Gate
-   Decides what information to discard from the cell state.
-   $f_t = \sigma(W_f \cdot [h_{t-1}, x_t] + b_f)$
-   Output close to 0 → forget; output close to 1 → keep.

#### 2. Input Gate
-   Decides what new information to store in the cell state.
-   $i_t = \sigma(W_i \cdot [h_{t-1}, x_t] + b_i)$
-   Candidate values: $\tilde{C}_t = \tanh(W_C \cdot [h_{t-1}, x_t] + b_C)$

#### 3. Output Gate
-   Decides what information to output from the cell state.
-   $o_t = \sigma(W_o \cdot [h_{t-1}, x_t] + b_o)$
-   Final output: $h_t = o_t * \tanh(C_t)$.

The **cell state** is updated at each time step, and the gates decide how information flows in and out.

#### Key Properties

-   **Handles Long-Term Dependencies**: LSTMs can remember information for many time steps, unlike basic RNNs.
-   **Mitigates Vanishing Gradient Problem**: The gating structure empowers the cell state to carry gradients directly through time.
-   **Flexible Memory Management**: Individual gates allow fine-grained control for overwriting, reading, and retaining information.

#### Common LSTM Variants

-   **Bidirectional LSTM (BiLSTM)**: Processes data from both past to future and future to past for richer context.

#### Applications

-   Language translation
-   Time-series forecasting
-   Video analysis
-   Speech and handwriting recognition
-   Sentiment analysis

### 3. Gated Recurrent Unit (GRU)

#### What is a GRU?

A **Gated Recurrent Unit (GRU)** is another improved version of the classic RNN, introduced to solve the vanishing gradient problem while making the architecture simpler than LSTM.

#### GRU Architecture

GRUs achieve memory management through two main gates:

#### 1. Update Gate ($z_t$)
-   Combines "forget" and "input" gates from LSTM into a single gate.
-   Controls the degree to which the unit updates its content.
-   $z_t = \sigma(W_z \cdot [h_{t-1}, x_t])$

#### 2. Reset Gate ($r_t$)
-   Controls how much of the past information to forget.
-   $r_t = \sigma(W_r \cdot [h_{t-1}, x_t])$

#### Candidate Activation & Final Output
-   Generate a candidate activation $\tilde{h}_t$ and combine with previous hidden state based on update gate.
-   $\tilde{h}_t = \tanh(W \cdot [r_t * h_{t-1}, x_t])$
-   Final hidden state: $h_t = (1 - z_t) * h_{t-1} + z_t * \tilde{h}_t$.

#### Key Properties

-   **Simpler Architecture**: Fewer parameters due to only two gates (update, reset), making GRUs computationally lighter than LSTMs.
-   **Fast Training**: Generally quicker to train and sometimes achieves similar or better performance as LSTM.
-   **Efficient for Long Dependencies**: Effectively manages long-term dependencies, addressing the vanishing gradient.

#### Differences: LSTM vs. GRU

| Feature           | LSTM                                                     | GRU                                                        |
| :---------------- | :------------------------------------------------------- | :--------------------------------------------------------- |
| Gates             | 3 (input, forget, output)                                | 2 (update, reset)                                          |
| Cell State        | Separate cell and hidden states                          | Merged cell and hidden state                               |
| Parameters        | More parameters, higher complexity/expressiveness        | Fewer parameters, lower computational cost                 |
| Use-case          | Preferred for complex, long sequences                    | Often preferred where computational efficiency is critical |

#### Applications

-   Language modeling
-   Machine translation
-   Sentiment analysis
-   Speech recognition

### 4. RNN, LSTM, GRU: Comparison Table

| Aspect                  | RNN                                                                  | LSTM                                                                               | GRU                                                        |
| :---------------------- | :------------------------------------------------------------------- | :--------------------------------------------------------------------------------- | :--------------------------------------------------------- |
| Memory                  | Short-term (limited by vanishing gradients)                          | Effective long-term through cell state/gates                                       | Effective long-term via simplified two-gate system         |
| Gates                   | None                                                                 | 3 (input, forget, output)                                                          | 2 (reset, update)                                          |
| Handling Long Sequence  | Poor (vanishing gradient)                                            | Good (mitigates vanishing/exploding gradients)                                     | Good (mitigates vanishing/exploding gradients)             |
| Speed                   | Fast to train but inaccurate for long dependencies                   | Slower due to complexity                                                           | Faster / fewer parameters                                  |
| Applications            | Text, speech, any sequence                                           | NLP, speech, time series, handwriting                                              | Most uses of LSTM, especially when speed is important      |

### 5. Architectures & Variants

-   **Deep RNNs**: Stack multiple recurrent layers for more expressive capabilities.
-   **Bidirectional RNNs**: Process input from both forward and backward directions for context.
-   **Sequence-to-Sequence RNNs**: Used in translation, summarization, etc.
-   **Hybrid Models**: Combine RNN/GRU/LSTM with CNNs for enhanced feature extraction.

### 6. Summary of Use Cases

-   **RNN**: Quick and efficient for short sequences or simpler tasks, but limited for long-range context.
-   **LSTM**: Best for long-sequence tasks with complex dependencies (e.g., translation, speech).
-   **GRU**: Similar to LSTM but faster and more efficient; great for marginal hardware or applications where speed matters.

### 7. Final Notes

-   All three architectures are foundational for modern deep learning on sequential data.
-   Advances like LSTM and GRU have allowed models to better capture long-range patterns critical in tasks from translation to stock prediction.
-   Choice between LSTM and GRU often depends on the specific use case, required complexity, and computational resources.

### Importing Important Libraries

```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout, Bidirectional, Input
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
```

### Tokenizing the words

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([story])
tokenizer.word_index
```

```python
input_sequences = []
for sentence in story.split('\n') :
    tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]

    for i in range(1,len(tokenized_sentence)):
        input_sequences.append(tokenized_sentence[:i+1])

print(input_sequences)
```

### Padding The Sequences

```python
max_len = max([len(x) for x in input_sequences])
padded_input_sequences = pad_sequences(input_sequences, maxlen = max_len, padding = 'pre')
padded_input_sequences
```

```python
X = padded_input_sequences[:,:-1]
y = padded_input_sequences[:,-1]
print(X.shape)
print(y.shape)
```

```python
y = to_categorical(y, num_classes = len(tokenizer.word_index) + 1)
print(y.shape)
```

```python
vocab_size = len(tokenizer.word_index) + 1
max_sequence_length = max_len - 1

model = Sequential([
    Input(shape=(max_sequence_length,)),
    Embedding(vocab_size, 150),
    Bidirectional(LSTM(256, return_sequences=True)),
    Dropout(0.3),
    LSTM(150),
    Dropout(0.3),
    Dense(vocab_size, activation='softmax')
])
```

```python
model.compile(loss = 'categorical_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
model.summary()
```

```python
model.fit(X, y, epochs = 100)
```

```python
text = input('Enter a text : ')
n = int(input('How many more words : '))
for i in range(n) :
    # tokenize
    token_text = tokenizer.texts_to_sequences([text])[0]
    # padding
    padded_token_text = pad_sequences([token_text], maxlen = 29, padding = 'pre')
    # predict
    pos = np.argmax(model.predict(padded_token_text))

    for word, index in tokenizer.word_index.items() :
        if index == pos :
            text = text + " " + word
            print(text)
```
