# OpenCV : Introduction

## 1. GUI Features

### `cv2.imread()`
Reads an image from a specified file path into a NumPy array.

* **Code:** `img = cv2.imread('path/to/image.jpg', cv2.IMREAD_COLOR)`
* **Syntax:** `cv2.imread(filename, flags)`
    * `filename`: **(str)** The full path to the image file.
    * `flags`: **(int)** Optional. Specifies how the image is loaded.
        * `cv2.IMREAD_COLOR` (or `1`): Loads a color image (default).
        * `cv2.IMREAD_GRAYSCALE` (or `0`): Loads a grayscale image.
        * `cv2.IMREAD_UNCHANGED` (or `-1`): Loads the image including alpha channel if present.
* **Customization:** Use different `flags` to load as grayscale or with transparency.

### `cv2.imshow()`
Displays an image in a designated window.

* **Code:** `cv2.imshow('My Image Window', img)`
* **Syntax:** `cv2.imshow(winname, mat)`
    * `winname`: **(str)** The name of the window in which the image is displayed. If a window with this name doesn't exist, it's created.
    * `mat`: **(numpy.ndarray)** The image (NumPy array) to be displayed.
* **Customization:** Choose any string for `winname`.

### `cv2.waitKey()`
Waits for a key press for a specified time. Essential for `imshow` to work correctly.

* **Code:** `key = cv2.waitKey(0)`
* **Syntax:** `cv2.waitKey(delay)`
    * `delay`: **(int)** Time in milliseconds to wait for a key event.
        * If `0` (or negative), it waits indefinitely until any key is pressed.
        * If positive (e.g., `1`), it waits for `1` ms and then continues. This is used in video loops.
* **Customization:** Set `delay` to `0` for static images or a small positive value for video/frame-by-frame processing. Returns the ASCII value of the pressed key.

### `cv2.destroyAllWindows()`
Destroys all open HighGUI windows created by OpenCV.

* **Code:** `cv2.destroyAllWindows()`
* **Syntax:** `cv2.destroyAllWindows()`
* **Customization:** None.

### `cv2.imwrite()`
Saves an image to a file.

* **Code:** `cv2.imwrite('output_image.png', img)`
* **Syntax:** `cv2.imwrite(filename, img)`
    * `filename`: **(str)** The full path and filename (including extension) to save the image. The extension dictates the image format (e.g., `.jpg`, `.png`, `.bmp`).
    * `img`: **(numpy.ndarray)** The image (NumPy array) to be saved.
* **Customization:** Change the `filename` and extension to save in different formats.

## 2. Core Operations

### `cv2.split()`
Splits a multi-channel image (like BGR) into individual single-channel images.

* **Code:** `b, g, r = cv2.split(img_color)`
* **Syntax:** `cv2.split(m)`
    * `m`: **(numpy.ndarray)** The multi-channel input image.
* **Customization:** None, it always returns a tuple of single-channel arrays in the order they appear in the input.

### `cv2.merge()`
Merges several single-channel images into a multi-channel image.

* **Code:** `img_merged = cv2.merge([b, g, r])`
* **Syntax:** `cv2.merge(mv)`
    * `mv`: **(list or tuple of numpy.ndarray)** A list or tuple of single-channel images to be merged. The order matters (e.g., `[b, g, r]` for a BGR image).
* **Customization:** Control the order of channels and which channels are included.

### `cv2.copyMakeBorder()`
Adds a border around an image.

* **Code:** `bordered_img = cv2.copyMakeBorder(img, 10, 10, 10, 10, cv2.BORDER_CONSTANT, value=[0, 0, 255])`
* **Syntax:** `cv2.copyMakeBorder(src, top, bottom, left, right, borderType, value)`
    * `src`: **(numpy.ndarray)** The input image.
    * `top, bottom, left, right`: **(int)** The width of the border in pixels for each side.
    * `borderType`: **(int)** Type of border to be added.
        * `cv2.BORDER_CONSTANT`: Adds a constant colored border (requires `value`).
        * `cv2.BORDER_REFLECT`: Border is formed by reflecting the image content across the border.
        * `cv2.BORDER_REPLICATE`: Repeats the last pixel from the original image at the border.
        * `cv2.BORDER_WRAP`: Wraps the image content around.
    * `value`: **(scalar or list/tuple)** Optional. Color of the border if `borderType` is `cv2.BORDER_CONSTANT`. For color images, it's a BGR tuple/list.
* **Customization:** Control border size, type, and color. 

### `cv2.add()`
Performs pixel-wise addition of two images (saturating arithmetic).

* **Code:** `added_img = cv2.add(img1, img2)`
* **Syntax:** `cv2.add(src1, src2)`
    * `src1, src2`: **(numpy.ndarray)** The two input images. Must have the same size and data type.
* **Customization:** NumPy's `img1 + img2` does modulo arithmetic, while `cv2.add` does saturating arithmetic (values cap at 255 for `uint8`), which is generally preferred for image addition.

### `cv2.addWeighted()`
Calculates the weighted sum of two images, useful for blending.

* **Code:** `blended_img = cv2.addWeighted(img1, 0.7, img2, 0.3, 0)`
* **Syntax:** `cv2.addWeighted(src1, alpha, src2, beta, gamma)`
    * `src1`: **(numpy.ndarray)** First input image.
    * `alpha`: **(float)** Weight of the first image.
    * `src2`: **(numpy.ndarray)** Second input image (same size and type as `src1`).
    * `beta`: **(float)** Weight of the second image.
    * `gamma`: **(float)** Scalar added to each sum (often 0).
* **Customization:** Adjust `alpha` and `beta` (usually `alpha + beta = 1`) to control the blending ratio. `gamma` can be used to adjust brightness.

## 3. Image Processing

### `cv2.cvtColor()`
Converts an image from one color space to another.

* **Code:** `gray_img = cv2.cvtColor(img_color, cv2.COLOR_BGR2GRAY)`
* **Syntax:** `cv2.cvtColor(src, code)`
    * `src`: **(numpy.ndarray)** The input image.
    * `code`: **(int)** Color space conversion code.
        * `cv2.COLOR_BGR2GRAY`: Converts BGR to grayscale.
        * `cv2.COLOR_BGR2HSV`: Converts BGR to HSV.
        * `cv2.COLOR_HSV2BGR`: Converts HSV back to BGR.
        * Many other codes available (e.g., `COLOR_BGR2LAB`, `COLOR_BGR2RGB`).
* **Customization:** Choose the appropriate `code` based on desired color space.

### `cv2.resize()`
Resizes an image to a specified dimension or scaling factor.

* **Code:** `resized_img = cv2.resize(img, (new_width, new_height), interpolation=cv2.INTER_LINEAR)`
* **Syntax:** `cv2.resize(src, dsize, fx=0, fy=0, interpolation=cv2.INTER_LINEAR)`
    * `src`: **(numpy.ndarray)** The input image.
    * `dsize`: **(tuple)** Desired output size `(width, height)`. If `(0,0)`, `fx` and `fy` are used.
    * `fx, fy`: **(float)** Scaling factors along the X and Y axes, respectively. Used if `dsize` is `(0,0)`.
    * `interpolation`: **(int)** Interpolation method.
        * `cv2.INTER_AREA`: Good for shrinking images (avoids moiré patterns).
        * `cv2.INTER_LINEAR`: Default. Good for zooming (bilinear interpolation).
        * `cv2.INTER_CUBIC`: Better quality for zooming but slower (bicubic interpolation).
* **Customization:** Specify exact `dsize` or scaling factors (`fx`, `fy`). Choose `interpolation` method for quality/speed trade-offs.

### `cv2.warpAffine()`
Applies an affine transformation (translation, rotation, scaling, shear) to an image.

* **Code:** `translated_img = cv2.warpAffine(img, M_translate, (cols, rows))`
* **Syntax:** `cv2.warpAffine(src, M, dsize)`
    * `src`: **(numpy.ndarray)** The input image.
    * `M`: **(numpy.ndarray)** The 2x3 transformation matrix.
    * `dsize`: **(tuple)** Size of the output image `(width, height)`.
* **Customization:** The transformation is entirely defined by the matrix `M`. You usually get this matrix from other functions:
    * `cv2.getRotationMatrix2D()` for rotation and scaling.
    * `cv2.getAffineTransform()` for transformations defined by 3 points.

### `cv2.getRotationMatrix2D()`
Calculates a 2x3 matrix for image rotation and scaling.

* **Code:** `M_rotate = cv2.getRotationMatrix2D((center_x, center_y), angle_deg, scale_factor)`
* **Syntax:** `cv2.getRotationMatrix2D(center, angle, scale)`
    * `center`: **(tuple)** The rotation center `(x, y)`.
    * `angle`: **(float)** Rotation angle in degrees. Positive values mean counter-clockwise rotation.
    * `scale`: **(float)** Isotropic scale factor.
* **Customization:** Define rotation center, angle, and scale.

### `cv2.getAffineTransform()`
Calculates a 2x3 affine transformation matrix from three pairs of corresponding points.

* **Code:** `M_affine = cv2.getAffineTransform(np.float32(src_pts), np.float32(dst_pts))`
* **Syntax:** `cv2.getAffineTransform(src, dst)`
    * `src`: **(numpy.ndarray)** Coordinates of 3 points in the input image. Must be `np.float32`.
    * `dst`: **(numpy.ndarray)** Coordinates of the corresponding 3 points in the output image. Must be `np.float32`.
* **Customization:** Define the transformation by specifying source and destination points.

### `cv2.warpPerspective()`
Applies a perspective transformation to an image.

* **Code:** `perspective_img = cv2.warpPerspective(img, M_perspective, (output_width, output_height))`
* **Syntax:** `cv2.warpPerspective(src, M, dsize)`
    * `src`: **(numpy.ndarray)** The input image.
    * `M`: **(numpy.ndarray)** The 3x3 perspective transformation matrix.
    * `dsize`: **(tuple)** Size of the output image `(width, height)`.
* **Customization:** The transformation is defined by `M`, usually obtained from `cv2.getPerspectiveTransform()`.

### `cv2.getPerspectiveTransform()`
Calculates a 3x3 perspective transformation matrix from four pairs of corresponding points.

* **Code:** `M_perspective = cv2.getPerspectiveTransform(np.float32(src_quad), np.float32(dst_quad))`
* **Syntax:** `cv2.getPerspectiveTransform(src, dst)`
    * `src`: **(numpy.ndarray)** Coordinates of 4 points in the input image. Must be `np.float32`.
    * `dst`: **(numpy.ndarray)** Coordinates of the corresponding 4 points in the output image. Must be `np.float32`.
* **Customization:** Define the transformation by specifying source and destination quadrilaterals.

## 4. Image Filtering (Smoothing/Blurring)

### `cv2.blur()`
Applies a normalized box filter (average blur) to an image.

* **Code:** `avg_blurred_img = cv2.blur(img, (5, 5))`
* **Syntax:** `cv2.blur(src, ksize)`
    * `src`: **(numpy.ndarray)** The input image.
    * `ksize`: **(tuple)** Blurring kernel size `(width, height)`. Both must be odd and positive.
* **Customization:** Adjust `ksize` for more or less blurring. Larger kernel = more blur.

### `cv2.GaussianBlur()`
Blurs an image using a Gaussian filter.

* **Code:** `gaussian_blurred_img = cv2.GaussianBlur(img, (5, 5), 0)`
* **Syntax:** `cv2.GaussianBlur(src, ksize, sigmaX, sigmaY=0, borderType=cv2.BORDER_DEFAULT)`
    * `src`: **(numpy.ndarray)** The input image.
    * `ksize`: **(tuple)** Gaussian kernel size `(width, height)`. Both must be odd and positive.
    * `sigmaX`: **(float)** Gaussian kernel standard deviation in X direction. If 0, it's calculated from `ksize`.
    * `sigmaY`: **(float)** Optional. Gaussian kernel standard deviation in Y direction. If 0, it's set to `sigmaX`.
* **Customization:** `ksize` controls blur amount. `sigmaX` (and `sigmaY`) provides more precise control over the blur intensity.

### `cv2.medianBlur()`
Applies a median filter, particularly effective for salt-and-pepper noise.

* **Code:** `median_blurred_img = cv2.medianBlur(img, 5)`
* **Syntax:** `cv2.medianBlur(src, ksize)`
    * `src`: **(numpy.ndarray)** The input image.
    * `ksize`: **(int)** Aperture linear size. Must be a single positive **odd** integer (e.g., 3, 5, 7...).
* **Customization:** Adjust `ksize` for blur strength.

### `cv2.bilateralFilter()`
Applies an edge-preserving, noise-reducing bilateral filter.

* **Code:** `bilateral_filtered_img = cv2.bilateralFilter(img, 9, 75, 75)`
* **Syntax:** `cv2.bilateralFilter(src, d, sigmaColor, sigmaSpace, borderType=cv2.BORDER_DEFAULT)`
    * `src`: **(numpy.ndarray)** The input image.
    * `d`: **(int)** Diameter of each pixel neighborhood.
    * `sigmaColor`: **(float)** Filter sigma in the color space. Larger value means more distant colors will be mixed.
    * `sigmaSpace`: **(float)** Filter sigma in the coordinate space. Larger value means more distant pixels will influence each other.
* **Customization:** `d` defines the neighborhood size. `sigmaColor` and `sigmaSpace` control the degree of smoothing based on color difference and spatial distance, respectively.

## 5. Morphological Transformations

### `cv2.erode()`
Erodes the boundaries of foreground objects (makes them smaller).

* **Code:** `eroded_img = cv2.erode(binary_img, kernel, iterations=1)`
* **Syntax:** `cv2.erode(src, kernel, anchor=(-1, -1), iterations=1, borderType=cv2.BORDER_DEFAULT, borderValue=cv2.morphologyDefaultBorderValue())`
    * `src`: **(numpy.ndarray)** The input image (typically binary).
    * `kernel`: **(numpy.ndarray)** The structuring element, usually from `cv2.getStructuringElement()`.
    * `iterations`: **(int)** Number of times erosion is applied.
* **Customization:** Adjust `kernel` size and shape, and `iterations` for stronger erosion.

### `cv2.dilate()`
Dilates the boundaries of foreground objects (makes them larger, fills small holes).

* **Code:** `dilated_img = cv2.dilate(binary_img, kernel, iterations=1)`
* **Syntax:** `cv2.dilate(src, kernel, anchor=(-1, -1), iterations=1, borderType=cv2.BORDER_DEFAULT, borderValue=cv2.morphologyDefaultBorderValue())`
    * `src`: **(numpy.ndarray)** The input image (typically binary).
    * `kernel`: **(numpy.ndarray)** The structuring element.
    * `iterations`: **(int)** Number of times dilation is applied.
* **Customization:** Adjust `kernel` size and shape, and `iterations` for stronger dilation.

## 6. Image Thresholding

### `cv2.threshold()`
Applies a fixed-level threshold to each pixel of a grayscale image.

* **Code:** `ret, binary_img = cv2.threshold(gray_img, 127, 255, cv2.THRESH_BINARY)`
* **Syntax:** `cv2.threshold(src, thresh, maxval, type)`
    * `src`: **(numpy.ndarray)** The input grayscale image.
    * `thresh`: **(float)** The threshold value.
    * `maxval`: **(float)** The maximum value to use with `cv2.THRESH_BINARY` and `cv2.THRESH_BINARY_INV`.
    * `type`: **(int)** Thresholding type.
        * `cv2.THRESH_BINARY`: `pixel > thresh` becomes `maxval`, else 0.
        * `cv2.THRESH_BINARY_INV`: `pixel > thresh` becomes 0, else `maxval`.
        * `cv2.THRESH_TRUNC`: `pixel > thresh` becomes `thresh`, else unchanged.
        * `cv2.THRESH_TOZERO`: `pixel > thresh` unchanged, else 0.
        * `cv2.THRESH_TOZERO_INV`: `pixel > thresh` becomes 0, else unchanged.
        * `cv2.THRESH_OTSU`: Special flag to automatically compute optimal threshold for bimodal images. Combine with other types using `+`, e.g., `cv2.THRESH_BINARY + cv2.THRESH_OTSU`.
    * `ret`: **(float)** The calculated threshold value (especially useful with Otsu's).
    * `dst`: **(numpy.ndarray)** The output binary image.
* **Customization:** Adjust `thresh`, `maxval`, and `type`. Use `cv2.THRESH_OTSU` for automatic thresholding.

### `cv2.adaptiveThreshold()`
Applies adaptive thresholding, where the threshold value is calculated for small regions.

* **Code:** `adaptive_thresh_img = cv2.adaptiveThreshold(gray_img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)`
* **Syntax:** `cv2.adaptiveThreshold(src, maxValue, adaptiveMethod, thresholdType, blockSize, C)`
    * `src`: **(numpy.ndarray)** The input grayscale image.
    * `maxValue`: **(float)** Non-zero value assigned to pixels above the threshold.
    * `adaptiveMethod`: **(int)** Adaptive thresholding method.
        * `cv2.ADAPTIVE_THRESH_MEAN_C`: Threshold is the mean of the neighborhood area.
        * `cv2.ADAPTIVE_THRESH_GAUSSIAN_C`: Threshold is a weighted Gaussian sum of the neighborhood.
    * `thresholdType`: **(int)** Type of thresholding to apply (`cv2.THRESH_BINARY` or `cv2.THRESH_BINARY_INV`).
    * `blockSize`: **(int)** Size of a pixel neighborhood that determines the threshold value. Must be a single **odd** integer (e.g., 3, 5, 7...).
    * `C`: **(float)** Constant subtracted from the mean or weighted sum.
* **Customization:** Adjust `adaptiveMethod`, `blockSize` (neighborhood size), and `C` to fine-tune the thresholding based on local image characteristics.

## 7. Edge Detection

### `cv2.Sobel()`
Calculates the first, second, third, or mixed image derivatives using an extended Sobel operator. Primarily used for edge detection.

* **Code:** `sobelx = cv2.Sobel(gray_img, cv2.CV_64F, 1, 0, ksize=5)`
* **Syntax:** `cv2.Sobel(src, ddepth, dx, dy, ksize=3, scale=1, delta=0, borderType=cv2.BORDER_DEFAULT)`
    * `src`: **(numpy.ndarray)** The input image.
    * `ddepth`: **(int)** Desired depth of the output image (e.g., `cv2.CV_64F` for float to handle negative values, then convert to `uint8` after taking absolute).
    * `dx`: **(int)** Order of the derivative in the x-direction.
    * `dy`: **(int)** Order of the derivative in the y-direction.
    * `ksize`: **(int)** Size of the extended Sobel kernel (must be 1, 3, 5, or 7).
* **Customization:** Choose derivative direction (`dx`, `dy`), `ddepth` (important for intermediate calculations), and `ksize`.

### `cv2.Laplacian()`
Calculates the Laplacian of an image, highlighting regions of rapid intensity change (edges).

* **Code:** `laplacian_img = cv2.Laplacian(gray_img, cv2.CV_64F)`
* **Syntax:** `cv2.Laplacian(src, ddepth, ksize=1, scale=1, delta=0, borderType=cv2.BORDER_DEFAULT)`
    * `src`: **(numpy.ndarray)** The input image.
    * `ddepth`: **(int)** Desired depth of the output image.
    * `ksize`: **(int)** Aperture size used to compute the second-derivative filters. Must be positive and odd. If 1, a 3x3 filter is used internally.
* **Customization:** Adjust `ddepth` and `ksize`.

### `cv2.Canny()`
Implements the multi-stage Canny edge detection algorithm, known for producing strong, thin edges.

* **Code:** `canny_edges = cv2.Canny(gray_img, 100, 200)`
* **Syntax:** `cv2.Canny(image, threshold1, threshold2, apertureSize=3, L2gradient=False)`
    * `image`: **(numpy.ndarray)** The 8-bit input image.
    * `threshold1`: **(float)** First threshold for the hysteresis procedure (lower threshold).
    * `threshold2`: **(float)** Second threshold for the hysteresis procedure (upper threshold).
    * `apertureSize`: **(int)** Optional. Aperture size for the Sobel operator (3, 5, or 7).
    * `L2gradient`: **(bool)** Optional. Boolean flag, `True` for more accurate L2 gradient magnitude calculation, `False` for L1.
* **Customization:** The two thresholds (`threshold1`, `threshold2`) are crucial for controlling edge sensitivity. Adjust `apertureSize` for different levels of detail in gradient calculation.

## 8. Histograms

### `cv2.calcHist()`
Calculates the histogram of an image or image region.

* **Code:** `hist = cv2.calcHist([gray_img], [0], None, [256], [0, 256])`
* **Syntax:** `cv2.calcHist(images, channels, mask, histSize, ranges)`
    * `images`: **(list of numpy.ndarray)** Source image(s) as a list (e.g., `[img]`).
    * `channels`: **(list of int)** Index of the channel(s) for which to calculate the histogram (e.g., `[0]` for grayscale/blue, `[1]` for green, `[2]` for red).
    * `mask`: **(numpy.ndarray)** Optional mask image (same size as input, `uint8`). If `None`, the whole image is used.
    * `histSize`: **(list of int)** Number of bins for each dimension.
    * `ranges`: **(list of float)** The lower and upper boundary of the pixel values to be binned.
* **Customization:** Calculate for specific channels, use a `mask` for an ROI, and define `histSize` (number of bins) and `ranges`.
