### What are LLM ?
**Large Language Models (LLMs)** are advanced AI systems that are designed to understand and generate human-like text and other content. They are a type of **foundation model** trained on a vast amount of data, which gives them the ability to perform a wide range of tasks, from writing code and translating languages to answering questions and summarizing documents.

### How LLMs Work

LLMs are built on the principles of **deep learning** and, as we've just discussed, are almost universally based on the **Transformer architecture**.

* **Massive Scale:** The "large" in LLM refers to two things: the **enormous datasets** they are trained on (trillions of words) and the huge number of **parameters** (billions or even trillions) within their neural networks. This scale is what allows them to capture the complex patterns and nuances of human language.
* **Transformer Architecture:** The core of an LLM is a Transformer. The self-attention mechanism allows these models to process entire sequences at once and understand the relationships between words, regardless of their distance in a sentence. This overcomes the limitations of older models like RNNs.
* **Training Objective:** LLMs are typically trained to predict the next word in a sentence. By doing this repeatedly on a massive corpus of text, they learn not just grammar and syntax, but also a remarkable amount of general knowledge about the world.

### Key Capabilities and Applications

Because of their training, LLMs have developed a number of powerful capabilities:

* **Text Generation:** They can create coherent and contextually relevant content, including articles, emails, creative writing, and marketing copy.
* **Question Answering:** LLMs can provide detailed answers to questions, often by synthesizing information from their training data.
* **Summarization:** They can condense long documents, articles, and research papers into concise summaries.
* **Language Translation:** LLMs can translate text between different languages with high accuracy and fluency.
* **Code Generation:** Many LLMs are trained on code and can generate, debug, and document code in various programming languages.

Popular examples of LLMs include OpenAI's GPT series, Google's Gemini, and others like BERT, Llama, and Claude. They are the driving force behind the recent surge in **generative AI**, making powerful AI tools accessible to the public through conversational interfaces like chatbots.

```python
import os
import google.generativeai as genai
from dotenv import load_dotenv
load_dotenv()

API_KEY = os.getenv("GEMINI_API_KEY")

genai.configure(api_key=API_KEY)

model = genai.GenerativeModel("gemini-1.5-flash-latest")

conversation_history = []

def get_gemini_response(prompt):
    global conversation_history
    try:
        chat_session = model.start_chat(history=conversation_history)
        response = chat_session.send_message(prompt)

        conversation_history.append({"role": "user", "parts": [prompt]})
        conversation_history.append({"role": "model", "parts": [response.text]})

        return response.text
    except Exception as e:
        return f"An error occurred: {e}"

def main_chat_loop():
    print("Welcome to the simple Gemini Chatbot!")
    print("Type 'exit' or 'quit' to end the conversation.")

    while True:
        user_input = input("You: ")

        if user_input.lower() in ['exit', 'quit']:
            print("Goodbye!")
            break

        print("Bot: Thinking...")
        bot_response = get_gemini_response(user_input)

        print(f"Bot: {bot_response}")

if __name__ == "__main__":
    main_chat_loop()
```

## 2. Prompts

```python
# 1. Dynamic and reusable Prompts
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate, FewShotPromptTemplate
prompt = PromptTemplate(template='Summarize {topic} in {emotion} tone')
print(prompt.format(topic = 'Cricket', emotion = 'fun'))
```

```python
# 2 Define the ChatPromptTemplate using form_template
chat_prompt = ChatPromptTemplate.from_messages([
    ('system', 'Hi you are a experienced {profession}'),
    ('user','tell me about {topic}'),
])
# Format the prompt with variable 
formated_messages = chat_prompt.format_messages(profession = 'Doctor', topic = 'Viral Fever')
print(formated_messages)
```

```python
# Few Shot Prompting
examples = [
    {'input': 'I was Charged twice for my subscription this month.', 'output': 'Billing Issue'},
    {'input': 'The App Crashes every time I try to login', 'output': 'Technical Problem'},
    {'input': 'Can You Explain How to Upgrade my Plans ?', 'output': 'General Inquiry'},
    {'input': 'I need a refund for payment I did not Authorize.', 'output': 'Billing Issue'},
]

# Create an example Template
example_template = '''
Ticket : {input}
Category : {output}
'''

# Build Few Shot Prompt Templates
few_shot_prompt = FewShotPromptTemplate(
    examples=examples,
    example_prompt=PromptTemplate(input_variables=['input', 'output'], template=example_template),
    prefix="Classify the following Customer Support Tickets into one of the categories : 'Billing Issue, Technical Problem or General Inquiry'. \n ",
    suffix='\nTicket : {user_input}\nCategory :',
    input_variables=['user_input'],
)

print(few_shot_prompt.format(user_input='My credit card payment was declined.'))
```

## 3. Chains

Chains are a way to combine different components, like prompt templates and Large Language Models (LLMs), into a single, predictable sequence to perform a specific task. They create a structured pipeline where the output of one step becomes the input for the next. This makes it easier to build complex applications that go beyond a single prompt-and-response cycle.

Here's a simple analogy: think of a chain as a pre-written recipe. You know exactly which ingredients to use (your components) and in what order to use them (the chain's sequence) to get a specific result (the final output).

## 5. Memory

* **ConversationBufferMemory:** Stores a transcript of recent messages. Great for short chats but can grow large quickly.
* **ConversationBufferWindowMemory:** Only keeps the last N interactions to avoid excessive token usage.
* **Summarizer-Based Memory:** Periodically summarizes older chat segments to keep a condensed memory footprint.
* **Custom Memory:** For advanced use cases, you can store specialized state (e.g., the user's preferences or key facts about them) in a custom memory class.

## 6. Agents

In the context of modern AI, an **AI agent** is an autonomous program that can understand complex goals, break them down into smaller steps, and use external tools and resources to execute those steps to achieve a final objective. Unlike a simple program that just responds to a single command, an AI agent can plan, reason, and act independently.

---

### How an AI Agent Works

An AI agent operates through a process of continuous action and feedback. 

* **Planning:** It takes a high-level goal and figures out the necessary sub-tasks.
* **Tool Use:** It identifies and uses appropriate tools to get specific information or perform actions. For instance, a software agent might use a web search tool to find information or a weather API to get a forecast.
* **Execution:** It carries out each step of its plan using the chosen tools.
* **Synthesis:** It compiles all the gathered information and results to present a complete and coherent solution to the user's initial request.

---

### Example

Imagine you ask a smart assistant, "Find a highly-rated Thai restaurant for dinner tonight and book a reservation for two at 7:00 PM."

An AI agent would:

1.  **Break down the goal:**
    * Find Thai restaurants near you.
    * Check their ratings to find a highly-rated one.
    * Check if that restaurant has a reservation available for two at 7:00 PM.
    * Make the reservation if possible.
2.  **Use tools:** It wouldn't know this information on its own. It would use a suite of tools like:
    * A **maps API** to find local restaurants.
    * A **review API** (like Yelp or Google Reviews) to check ratings.
    * A **booking API** (like OpenTable) to check availability and make the reservation.
3.  **Synthesize and act:** It would then confirm the reservation details with you and provide a confirmation.
