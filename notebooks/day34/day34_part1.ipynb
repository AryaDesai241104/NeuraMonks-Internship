## LangChain Day - 2 : Models

### Models
* Language Models
    * LLMs
    * Chat Models
* Embedding Models

LLMs (Large Language Models) are the foundational technology for generating text and performing a wide range of language tasks. Chat models are a **type of LLM** that has been specifically fine-tuned for conversational use.

In short:

* **LLMs** are general-purpose and can be used for tasks like summarization, translation, and code generation.
* **Chat models** are optimized to maintain context, coherence, and a natural flow in multi-turn conversations, making them ideal for chatbots and virtual assistants.

### Context Window
A **context window** is the maximum amount of text a Large Language Model (LLM) can consider at one time when generating a response. It's the model's short-term memory, encompassing the prompt and all previous conversational turns. The size of the context window, measured in **tokens**, determines how much information a model can remember and reference. For example, a model with a larger context window can summarize a longer document or maintain a coherent, multi-turn conversation without forgetting earlier details.

### Max Output Tokens
**Max output tokens** is a user-defined limit on the length of the response that the LLM generates. This setting is a hard cap that controls how many tokens the model is allowed to produce in a single output. It's distinct from the context window, as it only affects the output, not the input. For instance, if you set max output tokens to 50, the model will stop generating text after it has produced 50 tokens, even if it hasn't finished its thought or sentence. You can use this setting to manage costs and keep responses concise.

### Temperatures
Temperature is a setting that controls the **randomness** of a chat model's responses.

* **Low temperature** (e.g., 0.2) makes the model predictable and factual.
* **High temperature** (e.g., 0.8) makes it more creative and surprising.
