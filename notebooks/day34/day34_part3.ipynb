## Chat Models

### Closed Source Models

* **GPT series** (OpenAI): This includes models like **GPT-3.5** and **GPT-4**. These are the models that power popular services like ChatGPT and are known for their strong performance.
* **Gemini** (Google): This family of models, which includes **Gemini 1.5 Pro** and **Gemini 1.5 Flash**, is developed by Google and is used in a variety of its products and services.
* **Claude series** (Anthropic): Models such as **Claude 3 Opus**, **Claude 3 Sonnet**, and **Claude 3 Haiku** are developed by Anthropic and are notable for their focus on safety and constitutional AI principles.
* **Grok** (xAI): Developed by Elon Musk's xAI, Grok is the model that powers the chatbot on the X (formerly Twitter) platform.

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv('GEMINI_API_KEY')

model1 = ChatGoogleGenerativeAI(model="gemini-1.5-flash", google_api_key=api_key, temperature = 0, max_completion_tokens = 100)
result1 = model1.invoke('write a five line poem on cricket')

model2 = ChatGoogleGenerativeAI(model="gemini-1.5-flash", google_api_key=api_key, temperature = 1)
result2 = model2.invoke('write a five line poem on cricket')

print(result1.content)
print('\n')
print(result2.content)
```

### Local Installation

```python
from langchain_huggingface import ChatHuggingFace, HuggingFacePipeline
import os
from dotenv import load_dotenv

load_dotenv()

huggingface_token = os.environ.get("HUGGINGFACEHUB_ACCESS_TOKEN")

llm = HuggingFacePipeline.from_model_id(
    model_id = 'TinyLlama/TinyLlama-1.1B-Chat-v1.0',
    task = 'text-generation',
    pipeline_kwargs = dict(
        temperature = 0.5,
        max_new_tokens = 100
    )
)

model = ChatHuggingFace(llm=llm)
result = model.invoke('What is the capital of India ?')
print(result.content)
```

### Huggingface API Calling

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
import os
from dotenv import load_dotenv

load_dotenv()

huggingface_token = os.environ.get("HUGGINGFACEHUB_ACCESS_TOKEN")

llm = HuggingFaceEndpoint(
    repo_id='mistralai/Mistral-7B-Instruct-v0.3',
    task='text-generation',
    huggingfacehub_api_token=huggingface_token
)

model = ChatHuggingFace(llm=llm)
result = model.invoke('What is the capital of India ?')
print(result.content)
```
