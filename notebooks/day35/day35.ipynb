## LLM day-3 : Prompting

### 1. Chat Models and Message Types

A **Chat Model** is a type of Large Language Model (LLM) specifically trained to handle multi-turn conversations. Unlike a traditional text completion model that takes a single string as input and returns a single string, chat models understand the concept of a conversation history.

To communicate with a chat model, you don't just send a raw string. Instead, you use a list of **message objects** that provide a structured view of the conversation. There are three main types of messages:

* **`HumanMessage`**: Represents a message from the user. It's the standard way to provide new input to the model.
* **`AIMessage`**: Represents a message from the AI. This is how you provide context from previous AI responses to maintain the conversation's flow.
* **`SystemMessage`**: Provides high-level instructions or context for the AI's persona. For example, "You are a helpful assistant" sets the tone and purpose of the conversation.

By structuring the conversation this way, the model understands the roles of each participant and can generate more relevant and coherent responses.

***

### 2. Prompt Templates

**Prompt Templates** are pre-defined recipes for generating prompts. They allow you to create a reproducible and reusable structure for your prompts by defining placeholders for input variables. This prevents you from manually writing out the entire prompt each time and helps maintain consistency.

You've worked with two types of templates:

* **`PromptTemplate`**: Used for single-turn interactions where the input is a simple string. The template you created for the "Research Assistant" is a great example of this, as it takes inputs like `paper_input`, `style_input`, and `length_input` to construct a detailed request.
* **`ChatPromptTemplate`**: Designed specifically for chat models. It handles a list of messages, allowing you to define the roles (`system`, `human`, `ai`) and include placeholders for both a user's query and the chat history.

***

### 3. Managing Chat History and `MessagesPlaceholder`

In a chatbot, the **chat history** is a list of all past messages in the conversation. Sending this history with each new user query is crucial for the model to remember previous turns and maintain context. Without it, the model would treat every query as a brand new conversation.

The **`MessagesPlaceholder`** is a component within a `ChatPromptTemplate` that simplifies the inclusion of chat history. Instead of manually appending past messages to your prompt, you can use a placeholder (like `MessagesPlaceholder(variable_name="chat_history")`) that automatically inserts the entire list of previous `AIMessage` and `HumanMessage` objects into the correct position within the prompt.

***

### 4. The `invoke()` Method

The **`invoke()`** method is a fundamental part of the LangChain Expression Language (LCEL). It's the standard, synchronous way to execute a component, such as a language model, a prompt template, or an entire chain. When you call `model.invoke(prompt)`, you're telling LangChain to pass the formatted prompt to the specified model and wait for its response. It's a clean, consistent interface for running any chainable component in LangChain.
