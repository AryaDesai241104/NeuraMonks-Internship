### LLMS Ways to get Structured Outputs

### Using with_structured_output --> data format
1. TypedDict
2. Pydantic
3. json_schema

### TypedDict

**TypedDict** is a way to define a dictionary in Python where you specify what keys and values should exist. It helps ensure that your dictionary follows a specific structure.

**Why use TypedDict?**

* It tells Python what keys are required and what types of values they should have.
* It does not validate data at runtime (it just helps with type hints for better coding).

### Pydantic
**Pydantic** is data validation and data parsing library for python. It Ensures That the Data you work with is correct, structured, and type-safe

#### Basic Example
This feature lets you define a class with specific types for each field, like `str` or `int`. Pydantic then checks to make sure the data you provide matches those types.

#### Default Values
This means you can set a fallback value for a field. If you don't provide a value when creating the object, Pydantic will automatically use the default one you set.

#### Optional Fields
This allows a field to be either a specific type (like `str`) or `None`. It's a way to say a piece of data is not required.

#### Coerce
Pydantic automatically converts data to the correct type. For example, if you provide the string `'5'` for a field that expects an `int`, Pydantic will smartly convert it to the integer `5` without you having to do anything.

#### Built-in Validation
This is Pydantic's automatic checking for data integrity. It's the core of the library, ensuring that the data conforms to the rules you've defined, such as a field being a floating-point number.

#### Field Function
The `Field` function is an extra tool for adding more rules to a field. You can use it to set a `description`, add minimum or maximum value constraints, or even apply a regular expression (`regex`) to ensure a string matches a specific pattern.

#### Returns Pydantic Object
When you use Pydantic, it creates a special object. This object is easy to convert into other formats like a standard Python `dict` or a JSON string, which is useful for sending data over a network or saving it to a file.

### Use `TypedDict` if:
* You want to enforce a basic dictionary structure with type hints.
* You are not concerned with advanced data validation, such as checking if a number is positive.
* You trust the LLM to return data that conforms to the schema.
* You prefer a simple, built-in Python tool over an external library like Pydantic.

### Use Pydantic `BaseModel` if:
* You need robust data validation, like a field being a specific length or a number being within a certain range.
* You want automatic type conversion from a string to a number (e.g., `'100'` to `100`).
* You want to set default values for fields the LLM might miss.
* You prefer a type-safe object as the output, allowing you to use dot notation (e.g., `result.name`).

### Use JSON Schema if:
* You are working in an environment that already uses a standardized JSON format.
* You need to define a schema programmatically or dynamically.
* You want to define the validation rules in a format that's independent of any specific programming language.

### Using Output Parsers

1. StrOutputParser
2. JsonOutputParser
3. StructuredOutputParser
4. PydanticOutputParser

| Parser Type | Returns | Validation | Ease of Use | Best For |
| :--- | :--- | :--- | :--- | :--- |
| **StrOutputParser** | `str` | None | Very Easy | Debugging, getting raw text, or when no specific structure is needed. |
| **JsonOutputParser** | `dict` or `list` | Checks for valid JSON syntax. | Easy | Quick, simple structured data extraction without strict data type enforcement. |
| **PydanticOutputParser** | `Pydantic model object` | **Full validation** against a Pydantic schema, including type checking and custom rules. | Verbose | Building robust, type-safe, and production-ready applications with guaranteed output structures. |
| **StructuredOutputParser** | `dict` or `Pydantic model object` | Validates against a defined schema, but is often less comprehensive than Pydantic. | Medium | Defining named fields and simple structures, often serving as a more abstract way to handle structured output. |
