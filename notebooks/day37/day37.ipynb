## LangChain day - 5 : Chains & Runnables

In LangChain, **chains** are sequences of steps that work together to complete a task.
Each step can be an LLM call, a tool/API call, or some custom logic.
The output of one step is passed as the input to the next, creating a processing pipeline.
They allow you to combine multiple actions, like retrieving data, transforming it, and formatting the result.
There are different types, like `LLMChain` for a single prompt, or `SequentialChain` for multi-step workflows.
Chains make complex, multi-step AI applications easier to build, reuse, and manage.

1. Sequential Chains
2. Parallel Chains
3. Conditional Chains

### Sequential Chains

* **Definition:** This is the simplest type of chain, where the output of one step becomes the input for the next step.
* **Workflow:** The process flows in a straight line, one component after another.
* **Use Case:** Ideal for multi-step tasks where each step depends on the result of the previous one. A common example is taking a user's question, summarizing a document, and then answering the question based on the summary. 

### Parallel Chains

* **Definition:** This chain runs multiple sub-chains or components at the same time using the same initial input.
* **Workflow:** The initial input is simultaneously sent to all components, which process the data independently. The final output is a dictionary containing the results from all the sub-chains.
* **Use Case:** Best for tasks where you need to perform multiple, independent operations on the same input, such as extracting key topics, generating a summary, and analyzing the sentiment of a single text. This can significantly reduce latency compared to a sequential approach.

### Conditional Chains

* **Definition:** This chain introduces "if-else" logic into your workflow, allowing you to dynamically choose which sub-chain to execute based on a condition.
* **Workflow:** An initial component (like a sentiment analyzer or a classification model) processes the input. A condition is then evaluated based on this output. The program then branches and executes only the sub-chain that corresponds to the satisfied condition.
* **Use Case:** Perfect for building applications that need to adapt their behavior based on the input. For example, a customer support bot might route a user's feedback to a different department based on whether the sentiment is positive or negative.

## Runnables 

In LangChain, a **Runnable** is a foundational building block that represents a single, self-contained unit of work. The `Runnable` interface provides a standardized way to define, execute, and compose different components, such as models, prompt templates, and output parsers. This allows you to easily connect them into complex, production-ready applications.
\<br\>
### Key Characteristics of Runnables

  * **Standardized Interface:** All components that implement the `Runnable` interface share the same core methods, including `invoke` (for a single input), `batch` (for multiple inputs in parallel), and `stream` (for receiving output chunks as they're generated).
  * **Composability:** You can easily chain Runnables together using the pipe operator (`|`), creating powerful data pipelines where the output of one Runnable becomes the input of the next.
  * **Modularity:** They encapsulate specific logic, making your code easier to read, reuse, and debug. For example, a prompt template can be a Runnable, a language model can be a Runnable, and a combination of both is also a Runnable.
  * **Asynchronous Support:** Runnables are designed with asynchronous operations in mind, which allows for better performance and scalability, especially when dealing with I/O-bound tasks like making API calls to an LLM.

### Task-Specific Runnables
* **Definition:** These are core LangChain components that have been converted into Runnables so they can be used in pipelines.
* **Purpose:** Perform task-specific operations like LLM calls, prompting, retrieval, etc.
* **Examples:**
    * `ChatOpenAI`: Runs an LLM model.
    * `PromptTemplate`: Formats prompts dynamically.
    * `Retriever`: Retrieves relevant documents.

### Runnable Primitives
* **Definition:** These are fundamental building blocks for structuring execution.
* **Purpose:** They help orchestrate execution by defining how different Runnables flow (sequentially, in parallel, conditionally, etc.).
* **Examples:**
    * `RunnableSequence`: Runs steps in order (`|` operator).
    * `RunnableParallel`: Runs multiple steps simultaneously.
    * `RunnableMap`: Maps the same input across multiple functions.
    * `RunnableBranch`: Implements conditional execution (if-else logic).
    * `RunnableLambda`: Wraps custom Python functions into Runnables.
    * `RunnablePassthrough`: Just forwards input as output (acts as a placeholder).
